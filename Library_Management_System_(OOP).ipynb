```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_issued = False

class Library:
    def __init__(self):
        self.books = []

    def add_book(self, title, author):
        new_book = Book(title, author)
        self.books.append(new_book)
        print(f"Added: '{title}' by {author}")

    def remove_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                self.books.remove(book)
                print(f"Removed: '{book.title}'")
                return
        print(f"Book '{title}' not found in the library.")

    def issue_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if not book.is_issued:
                    book.is_issued = True
                    print(f"Issued: '{book.title}'")
                else:
                    print(f"Sorry, '{book.title}' is already issued.")
                return
        print(f"Book '{title}' not found in the library.")

    def return_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if book.is_issued:
                    book.is_issued = False
                    print(f"Returned: '{book.title}'")
                else:
                    print(f"'{book.title}' was not issued.")
                return
        print(f"Book '{title}' does not belong to this library.")

    def display_books(self):
        print("\n--- Library Catalog ---")
        for book in self.books:
            status = "Issued" if book.is_issued else "Available"
            print(f"- '{book.title}' by {book.author} [{status}]")
        print("-----------------------\n")

# --- Example Usage ---
lib = Library()
lib.add_book("Clean Code", "Robert C. Martin")
lib.add_book("The Pragmatic Programmer", "Andrew Hunt")
lib.display_books()

lib.issue_book("Clean Code")
lib.issue_book("Clean Code")  # Already issued check
lib.return_book("Clean Code")
lib.remove_book("The Pragmatic Programmer")
lib.display_books()
```
