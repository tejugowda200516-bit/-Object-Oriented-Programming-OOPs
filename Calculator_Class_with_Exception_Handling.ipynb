```python
class Calculator:
    def _validate_inputs(self, a, b):
        if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
            raise TypeError("Both inputs must be numbers (int or float).")

    def add(self, a, b):
        try:
            self._validate_inputs(a, b)
            return a + b
        except TypeError as e:
            return f"Error: {e}"

    def subtract(self, a, b):
        try:
            self._validate_inputs(a, b)
            return a - b
        except TypeError as e:
            return f"Error: {e}"

    def multiply(self, a, b):
        try:
            self._validate_inputs(a, b)
            return a * b
        except TypeError as e:
            return f"Error: {e}"

    def divide(self, a, b):
        try:
            self._validate_inputs(a, b)
            return a / b
        except ZeroDivisionError:
            return "Error: Division by zero is not allowed."
        except TypeError as e:
            return f"Error: {e}"

# --- Example Usage ---
calc = Calculator()
print("Addition:", calc.add(10, 5))
print("Subtraction:", calc.subtract(10, 5))
print("Multiplication:", calc.multiply(10, 5))
print("Division:", calc.divide(10, 2))
print("Divide by Zero:", calc.divide(10, 0))       # Handled ZeroDivisionError
print("Invalid Input:", calc.add(10, "five"))      # Handled TypeError
```
